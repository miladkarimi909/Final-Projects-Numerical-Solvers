```python
# Sum/Product + Comparison (Colab)

def get_number(prompt):
    while True:
        s = input(prompt).strip()
        try:
            return float(s)
        except ValueError:
            print("Input must be numeric. Try again.")

A = get_number("Enter number A: ")
B = get_number("Enter number B: ")

s = A + B
p = A * B

print(f"Sum (A+B) = {s}")
print(f"Product (A*B) = {p}")

if s > p:
    print("Condition 1: The SUM is greater than the PRODUCT.")
elif s < p:
    print("Condition 2: The SUM is less than the PRODUCT.")
else:
    print("The SUM equals the PRODUCT.")

```
